## Task 1: Regex Fundamentals (Character Classes and Quantifiers)

Extract digits, capitalized words, 4 digit years, and words 3 to 6 letters long.


### a to d. Regular expressions

```python
text = "Ayesha studies NLP in Islamabad in 2026. Python 3 supports AI and BERT. Code 1234!"
print("All digits:", re.findall(r"\d", text))
print("Capitalized words:", re.findall(r"\b[A-Z][a-zA-Z]*\b", text))
print("Exactly 4 digit numbers:", re.findall(r"(?<!\d)\d{4}(?!\d)", text))
print("Words containing 3 to 6 letters:", re.findall(r"\b[A-Za-z]{3,6}\b", text))
```

## Task 2: Pattern Based Extraction (Real World Data)

Identify common real world information from a mixed paragraph.


### a to d. Extract emails, phones, dates and percentages

```python
text = """Email ayesha@example.com or help@uni.edu. Call 0301-1234567 or
0333-7654321. Orders AB-204 and ZX-450 were placed on 21/10/2026.
Results were 45% and 12.5%."""
print("Emails:", re.findall(r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}", text))
print("Phones:", re.findall(r"\b03\d{2}-\d{7}\b", text))
print("Dates:", re.findall(r"\b\d{2}/\d{2}/\d{4}\b", text))
print("Percentages:", re.findall(r"\b\d+(?:\.\d+)?%", text))
```

## Task 3: Word Tokenization (NLTK vs spaCy)

Use NLTK word_tokenize and a blank English spaCy tokenizer.


### a to d. Compare the token lists

```python
paragraph = ("I don't think it's difficult. Ayesha can't attend today. "
             "We're studying language models! Isn't this interesting?")
# preserve_line=True avoids needing to download the sentence tokenizer.
nltk_tokens = nltk.word_tokenize(paragraph, preserve_line=True)
nlp_blank = spacy.blank("en")
spacy_tokens = [token.text for token in nlp_blank(paragraph)]
print("NLTK:", nltk_tokens)
print("spaCy:", spacy_tokens)
print("NLTK token count:", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))
# Observation 1: NLTK and spaCy can split contractions differently.
# Observation 2: Punctuation and apostrophes may lead to different token counts.
print("NLTK only:", [x for x in nltk_tokens if x not in spacy_tokens])
print("spaCy only:", [x for x in spacy_tokens if x not in nltk_tokens])
```

## Task 4: Token Attributes (Categorizing Tokens)

Inspect basic token attributes.


### a to d. Token categories

```python
sentence = "Ayesha paid $50 for 2 books. Email ayesha@example.com today."
doc = nlp_blank(sentence)
print(f"{'Token':<23} {'is_alpha':<10} {'like_num':<10} {'like_email':<12}")
for token in doc:
    print(f"{token.text:<23} {str(token.is_alpha):<10} {str(token.like_num):<10} {str(token.like_email):<12}")
currency = [token.text for token in doc if token.is_currency]
print("Currency tokens:", currency)
```

## Task 5: Sentence Segmentation (Handling Tricky Cases)

Check abbreviations such as Dr. and Mr. in sentence segmentation.


### a to d. Segment sentences and evaluate abbreviation handling

```python
text = "Dr. Sara met Mr. Ali at COMSATS. They discussed NLP. It was a useful meeting."
try:
    trained_nlp = spacy.load("en_core_web_sm")
    model_type = "Trained en_core_web_sm model"
except OSError:
    # Fallback keeps the notebook runnable, but is NOT a trained model.
    trained_nlp = spacy.blank("en")
    trained_nlp.add_pipe("sentencizer")
    model_type = "Fallback rule-based sentencizer (install en_core_web_sm for the required trained model)"

sentences = [sent.text.strip() for sent in trained_nlp(text).sents]
print("Sentence segmenter:", model_type)
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
print("Total sentences:", len(sentences))
# The intended result is 3 sentences; splitting after 'Dr.' or 'Mr.' is incorrect.
print("Abbreviations handled correctly:", len(sentences) == 3 and sentences[0].startswith("Dr. Sara met Mr. Ali"))
```
