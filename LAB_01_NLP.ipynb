# Task 07: String Validation Methods

```python
# (a)  Given text = "NLP is the future of AI research", use startswith() to check whether the text begins with "NLP", and endswith() to check whether it ends with "research". Print a clear message for each result. Then use find() to get the position of the word "future" in the string.
text = "NLP is the future of AI research"
if text.startswith("NLP"):
    print("The text starts with NLP.")
else:
    print("The text does not start with NLP.")
if text.endswith("research"):
    print("The text ends with research.")
else:
    print("The text does not end with research.")
print(text.find("future"))
# (b)  Create a mixed list: tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]. Loop through the list and for each token print whether it is purely alphabetic (isalpha()), purely numeric (isdigit()), or mixed. Label each output clearly.
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for t in tokens:
    if t.isalpha():
        print(f"{t}: purely alphabetic")
    elif t.isdigit():
        print(f"{t}: purely numeric")
    else:
        print(f"{t}: mixed")
# (c)  Given the list words = ["running", "jumped", "NLP", "ai2025", "data", "42"]. Use a loop with isalpha() and isdigit() to separate the list into three groups: pure_words, pure_numbers, and mixed_tokens. Print each group.
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print(pure_words)
print(pure_numbers)
print(mixed_tokens)
# (d)  Write a function called validate_token(token) that returns the string "WORD" if the token is purely alphabetic, "NUMBER" if it is purely numeric, and "MIXED" otherwise. Apply this function to each element in the list ["chat", "99", "gpt4", "bert", "2024", "nlp!"] using a loop and print the token alongside its label.
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

for t in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(t, "->", validate_token(t))
```

# Task 08: Functions and Lambda for NLP Pipelines

```python
# (a)  Write a regular function called clean(text) that takes a string, strips whitespace, converts it to lowercase, and returns the result. Then write an equivalent lambda function called clean_lambda that performs the same two operations in one line. Test both on the string "  HELLO Natural Language Processing  " and print their outputs.
def clean(text):
    text = text.strip()
    text = text.lower()
    return text
clean_lambda = lambda text: text.strip().lower()

test = "  HELLO Natural Language Processing  "
print(clean(test))
print(clean_lambda(test))
# (b)  Write a function called tokenize(text) that takes a cleaned string and returns a list of individual word tokens using split(). Then write another function called remove_short(tokens, n) that takes the token list and an integer n, and returns only tokens whose length is greater than n. Chain both functions on the sentence "I love working with NLP tools in Python daily" with n = 3 and print the result.
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
result = remove_short(tokenize(sentence), 3)
print(result)
# (c)  Write a function called word_lengths(tokens) that takes a list of tokens and returns a dictionary where each key is a word and each value is the length of that word. Call the function on tokens = ["language", "model", "ai", "processing", "deep"] and print the resulting dictionary.
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print(word_lengths(tokens))
# (d)  Combine the functions from parts (b) and (c) into a full pipeline function called nlp_pipeline(text, min_len) that: cleans the text, tokenises it, removes short words, and returns a length dictionary. Call this pipeline on two different sentences and print the outputs.
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

print(nlp_pipeline("  Deep Learning Powers Modern AI Systems  ", 3))
print(nlp_pipeline("Natural Language Processing is a fascinating field", 4))
```

# Task 09: map(), filter(), sorted() and Counter

```python
# (a)  Given words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]. Use map() with a lambda to: (i) convert every word to lowercase — print the result as a list, and (ii) compute the character length of each word — print the result as a list.
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]
lower_words = list(map(lambda w: w.lower(), words))
print(lower_words)
lengths = list(map(lambda w: len(w), words))
print(lengths)
# (b)  Using the same words list, apply filter() with a lambda to create two filtered lists: (i) words_long — only words with more than 3 characters, and (ii) words_with_t — only words containing the letter "t" (case-insensitive). Print both lists.
words_long = list(filter(lambda w: len(w) > 3, words))
print(words_long)

words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print(words_with_t)
# (c)  Sort the words list using sorted() in three different ways and print each result with a label: (i) alphabetically (default order), (ii) by word length ascending using key=len, (iii) by word length descending using reverse=True.
alpha_sorted = sorted(words)
print("Alphabetical:", alpha_sorted)

len_sorted_asc = sorted(words, key=len)
print("By length (ascending):", len_sorted_asc)

len_sorted_desc = sorted(words, key=len, reverse=True)
print("By length (descending):", len_sorted_desc)
# (d)  Import Counter from collections. Given the sentence "nlp nlp python ai nlp python text ai nlp python text", split it into words and create a Counter called freq. Print the full counter, then print: (i) the 3 most common words using most_common(), (ii) the frequency of the word "python", and (iii) the total number of words and the size of the unique vocabulary.
from collections import Counter
sentence = "nlp nlp python ai nlp python text ai nlp python text"
tokens = sentence.split()
freq = Counter(tokens)

print(freq)
print(freq.most_common(3))
print(freq["python"])
print(sum(freq.values()))
print(len(freq))
```
